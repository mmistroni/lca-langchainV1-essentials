```python
from typing import Annotated
import operator
from typing_extensions import TypedDict
from langgraph.graph import StateGraph, START, END
from langchain.agents import create_agent
from langchain_google_genai import ChatGoogleGenerativeAI # Ensure this is installed
from langchain_core.output_parsers import StrOutputParser
from langchain_classic.output_parsers import DatetimeOutputParser
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from langchain_core.prompts import (SystemMessagePromptTemplate, 
                                    HumanMessagePromptTemplate, 
                                    AIMessagePromptTemplate, 
                                    ChatPromptTemplate,
                                    FewShotChatMessagePromptTemplate)

```

```python
model = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite", 
                               max_tokens=100,
                               temperature=0,
                               model_kwargs={'seed' :365}
                               )
```

```python
message_h = HumanMessage(content = f'''
     When was the Danish poet Piet Hein born?
     {DatetimeOutputParser().get_format_instructions()}
     
     '''
     )
response = model.invoke([message_h])
#print(response.content)
```

```python
date_output_parser = DatetimeOutputParser()
response_parsed = date_output_parser.invoke(response)
response_parsed
```
