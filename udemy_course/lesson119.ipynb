```python
from typing import Annotated
import operator
from typing_extensions import TypedDict
from langgraph.graph import StateGraph, START, END
from langchain.agents import create_agent
from langchain_google_genai import ChatGoogleGenerativeAI # Ensure this is installed
print('foo')

```

```python
model = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite", messages=[{"role": "system", "content": "You are a helpful assistant."}])

conversation_history = []

while True:
    user_input = input("You: ")
    conversation_history.append(user_input)
    if user_input.lower() == "exit":
        break
    response = model.invoke(conversation_history)
    conversation_history.append(response.text)
    print(f"AI: {response.text}")

print(conversation_history)
```
